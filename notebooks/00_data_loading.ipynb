## Cookie Cats A/B test

One row per player: group (`gate_30` or `gate_40`), rounds played during the first 14 days after install, return on day 1 and day 7.

```python
con = connect()
con.sql("DESCRIBE cookie_cats").show()
con.sql("SELECT * FROM cookie_cats LIMIT 5").show()
```

## Player behaviour dataset

One row per player: sessions, playtime, level, purchases and an engagement label. Used for the segmentation check (notebook 03).

```python
con.sql("DESCRIBE gaming_behavior").show(max_rows=20)
con.sql("SELECT * FROM gaming_behavior LIMIT 5").show(max_width=200)
con.close()
```
